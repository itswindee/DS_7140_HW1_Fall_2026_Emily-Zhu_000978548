## 1.	Python Basics 

The Museum of Modern Art (MoMA) charges different admission prices depending on the visitor's age and student status. The admission prices are shown below.

Catergory and Admission Price: 
Adult (Age 17-64): USD 30;
Full-time Student: USD 17;
Senior (Age 65+): USD 22;
Age 16 or younger: Free.

1.1.	Define a Python function named ticket_price(age, is_student) that calculates and returns the admission price for one visitor based on the visitor's age and student status. (2 pts)

```python
def ticket_price(age, is_student):
    if age <= 16:
        return 0
    elif is_student:
        return 17
    elif age >= 65:
        return 22
    elif 17 <= age <= 64:
        return 30
    else:
        return 0
```

1.2.	Using the function developed in Question 1.1, determine the admission price for each of the following visitors:

a.	A 25-year-old adult who is not a student. (0.5 pt)

b.	A 70-year-old senior who is not a student. (0.5 pt)

c.	An 18-year-old full-time student. (0.5 pt)


```python
# a. A 25-year-old adult who is not a student
price_a = ticket_price(25, False)
print(f"a. 25-year-old non-student: ${price_a}")

# b. A 70-year-old senior who is not a student
price_b = ticket_price(70, False)
print(f"b. 70-year-old non-student: ${price_b}")

# c. An 18-year-old full-time student
price_c = ticket_price(18, True)
print(f"c. 18-year-old full-time student: ${price_c}")
```

1.3.	A family of three plans to visit MoMA. Their information is listed below:

Visitor 1: age 45, not a student

Visitor 2: age 40, not a student

Visitor 3: age 12, full-time student

Visitor 4: age 22, full-time student

Using the function developed in Question 1.1, calculate and report the total admission price for the family. (1.5 pts)

Hint: The function parameter “is_student” should take either True or False. For example, (age = 18, is_student = False) represents a person aged 18 but not in a student status.

```python
# Store the visitor information (each item is a tuple: (age, is_student))
visitors = [
    (45, False),  # Visitor 1: age 45, not a student
    (40, False),  # Visitor 2: age 40, not a student
    (12, True),   # Visitor 3: age 12, full-time student
    (22, True)    # Visitor 4: age 22, full-time student
]

# Calculate the total admission price
total_price = sum(ticket_price(age, is_student) for age, is_student in visitors)

# Report the result
print(f"Total admission price: ${total_price}")
```

# 2.NumPy 

Create a 10 x 10 matrix (2D-array) using the code below.

```python
%pip install numpy
import numpy as np

np.random.seed(42)
A = np.random.randint(1, 101, size=(10, 10))
```

Please complete the following tasks:

2.1. Sort the values in each row of A in ascending order. Store the result in a new array named A_sorted and print the first three rows of A_sorted. (1 pt)

```python
A_sorted = np.sort(A, axis=1)
print("First three rows of A_sorted:")
print(A_sorted[:3])
```

2.2. Reshape the original matrix A from a 10 × 10 array into a 20 × 5 array. Store the reshaped array in a variable named A_reshaped. Print the resulting matrix and its shape. (1 pt)

```python
A_reshaped = A.reshape(20, 5)
print("\nA_reshaped:")
print(A_reshaped)
print("Shape of A_reshaped:", A_reshaped.shape)
```

2.3. Using array slicing, extract the elements located in rows 3 through 6 and columns 4 through 8 of A. Store the result in a new array and print it. (1 pt)

```python
A_sliced = A[3:7, 4:9]
print("Extracted array: \n", A_sliced)
```

2.4. Split the original matrix A into two equal parts vertically, so that each resulting array contains 5 rows and 10 columns. Print the rank and size of each resulting array. (1 pt)

```python
top_half, bottom_half = np.split(A, 2, axis=0)  # or np.vsplit(A, 2)

print("\nTop half rank (ndim):", top_half.ndim, "| size:", top_half.size)
print("Bottom half rank (ndim):", bottom_half.ndim, "| size:", bottom_half.size)
```

2.5. Using the two arrays obtained in Question 2.4, concatenate them side by side to create a new array with 5 rows and 20 columns. Print the resulting array and its shape. (1 pt)

```python
A_concatenated = np.hstack((top_half, bottom_half))  # or np.concatenate((top_half, bottom_half), axis=1)

print("\nConcatenated array:")
print(A_concatenated)
print("Shape of concatenated array:", A_concatenated.shape)
```
